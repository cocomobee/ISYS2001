## Your final prompt

> Paste your complete, finished prompt below (double-click to edit).

```
Could you build a python program for a simple purchase advisor.
The purchase advisor will take the inputs of how much the person's yearly income is, how much money they have and have expensive their purchase is.
It will then output to the user if it recommends for or against the purchase.
The decision logic will be if it is more than 50% of their current money and equal to or less than 10% of their yearly income it is recommended, 50% or less of their current money and 25% or less of their yearly income it is recommended.
More than 25% of their yearly income but equal to or less than 25% of their current money is also recommended. Else not recommend.
```

````python
# Paste the code your prompt produced here, then run it to show it works.
yearly_income = float(input("Enter your yearly income: $"))
current_money = float(input("Enter how much money you currently have: $"))
purchase_cost = float(input("Enter the cost of the purchase: $"))

if purchase_cost > current_money * 0.50 and purchase_cost <= yearly_income * 0.10:
    recommendation = "Recommended"

elif purchase_cost <= current_money * 0.50 and purchase_cost <= yearly_income * 0.25:
    recommendation = "Recommended"

elif purchase_cost > yearly_income * 0.25 and purchase_cost <= current_money * 0.25:
    recommendation = "Recommended"

else:
    recommendation = "Not recommended"

print("Purchase advice:", recommendation)
````
